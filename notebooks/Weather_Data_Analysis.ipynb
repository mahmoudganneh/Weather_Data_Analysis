```python
df['Date'] = pd.to_datetime(df['Date'])
```

```python
df.info()
```

```python
df = df.sort_values('Date')
df.head()
```

```python
df['Year'] = df['Date'].dt.year
df['Month'] = df['Date'].dt.month
df['Month_Name'] = df['Date'].dt.month_name()
df['Day'] = df['Date'].dt.day
```

```python
def season(month):
    if month in [12,1,2]:
        return "Winter"
    elif month in [3,4,5]:
        return "Summer"
    elif month in [6,7,8,9]:
        return "Monsoon"
    else:
        return "Post-Monsoon"

df["Season"] = df["Month"].apply(season)
```

```python
plt.figure(figsize=(15,6))
plt.plot(df["Date"], df["Temperature_Avg (°C)"], color="red")
plt.title("Daily Average Temperature Trend")
plt.xlabel("Date")
plt.ylabel("Temperature (°C)")
plt.grid(True)

plt.savefig("plots/temp_trend.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
plt.figure(figsize=(15,6))
plt.plot(df["Date"], df["Rainfall (mm)"], color="blue")
plt.title("Daily Rainfall Trend")
plt.xlabel("Date")
plt.ylabel("Rainfall (mm)")
plt.grid(True)

plt.savefig("plots/rainfall_trend.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
monthly_temp = df.groupby("Month_Name")["Temperature_Avg (°C)"].mean()

month_order = ["January","February","March","April","May","June",
               "July","August","September","October","November","December"]

monthly_temp = monthly_temp.reindex(month_order)

plt.figure(figsize=(12,6))
monthly_temp.plot(kind="bar", color="tomato")

plt.title("Monthly Average Temperature")
plt.xlabel("Month")
plt.ylabel("Average Temperature (°C)")

plt.savefig("plots/monthly_avg_temperature.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
monthly_rain = df.groupby("Month_Name")["Rainfall (mm)"].mean()
monthly_rain = monthly_rain.reindex(month_order)

plt.figure(figsize=(12,6))
monthly_rain.plot(kind="bar", color="royalblue")

plt.title("Monthly Average Rainfall")
plt.xlabel("Month")
plt.ylabel("Rainfall (mm)")

plt.savefig("plots/monthly_rainfall.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
plt.figure(figsize=(12,8))

corr = df.select_dtypes(include="number").corr()

sns.heatmap(corr,
            annot=True,
            cmap="coolwarm",
            fmt=".2f")

plt.title("Correlation Heatmap")

plt.savefig("plots/correlation_heatmap.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
df["Rolling_Temp"] = df["Temperature_Avg (°C)"].rolling(7).mean()

plt.figure(figsize=(15,6))

plt.plot(df["Date"],
         df["Temperature_Avg (°C)"],
         alpha=0.4,
         label="Daily")

plt.plot(df["Date"],
         df["Rolling_Temp"],
         linewidth=3,
         label="7-Day Rolling Average")

plt.legend()

plt.title("7-Day Rolling Average Temperature")

plt.savefig("plots/rolling_average_temperature.png",
            dpi=300,
            bbox_inches="tight")

plt.show()
```
